```python
import seaborn as sns
df = sns.load_dataset('tips')
```

```python
# 막대 그래프
sns.barplot(x=df['size'], y=df['total_bill'], hue=df['time'])
```

```python
# 막대 그래프
sns.countplot(x=df['size'], hue=df['time'])
```

```python
# 산점도(산포도)
sns.scatterplot(x=df['tip'], y=df['total_bill'], hue=df['time'])
```
