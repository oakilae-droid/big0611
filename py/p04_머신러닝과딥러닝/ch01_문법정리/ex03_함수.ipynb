# 사용자 정의 함수
## 함수 정의
def 함수명([매개변수1, 매개변수2, ...]): 문장 [return 반환값]

## 함수 호출
함수명([인자1, 인자2, ...])

```python
# 사칙연산 계산기
def calculator(x, y, op):
    err = 0
    result = 0

    try:
        if op == '+':
            result = x + y
        elif op == '-':
            result = x - y
        elif op == '*':
            result = x * y
        elif op == '/':
            # if y != 0:
            result = x / y
        else:
            err = -1
            print('계산 실패')

        return err, result
    except:
        print('에러 발생')
        err = -1
        return err, result
```

```python
# 함수 호출
calculator(1, 2, '+')
```

```python
calculator(5, 2, '?')
```

```python
calculator('5', '2', '/')
```

```python
calculator(5, 0, '/')
```
