## 2. 반복문
### (1) for: 반복 횟수가 정해져 있을 때
반복자: 리스트, 튜플, 세트, 딕셔너리, range()

    문장

### (2) while: 반복 횟수x, 예) 무한 반복
while (True):
    문장

```python
numbers = range(10)

for i in numbers:
    if i < 5:
        print('{}는 5보다 작습니다.'.format(i))
    elif i ==5:
        print('{}는 5보다 같습니다.'.format(i))
    else:
        print('{}는 5보다 급습니다.'.format(i))
        
```
