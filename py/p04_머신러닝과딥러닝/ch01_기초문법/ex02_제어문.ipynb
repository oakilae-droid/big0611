```python
# 예시: 고객의 구매 금액에 따른 등급 분류
purchase_amount = 120000

if purchase_amount >= 100000:
    print("VIP 회원입니다.")
elif purchase_amount >= 50000:
    print("우수 회원입니다.")
else:
    print("일반 회원입니다.")
```

```python
# 예시: 5명의 시험 점수 리스트에서 80점 이상인 사람만 세기
scores = [75, 92, 85, 60, 95]
pass_count = 0

for score in scores:
    if score >= 80:  # 반복문 안에 조건문을 넣을 수 있습니다!
        pass_count = pass_count + 1

print(f"합격자 수는 {pass_count}명입니다.")
```

```python
# 예시: 배터리가 10% 이상일 때만 데이터 수집 진행
battery = 15

while battery > 10:
    print("데이터를 수집 중입니다...")
    battery = battery - 2  # 배터리 감소 표현
print("배터리가 부족하여 수집을 중단합니다.")
```

```python
# 기존 방식: 0부터 9까지 숫자 중 짝수만 골라 리스트 만들기
even_numbers = []
for i in range(10):
    if i % 2 == 0:
        even_numbers.append(i)

# 한 줄로 줄이기 (결과는 동일)
even_numbers = [i for i in range(10) if i % 2 == 0]
print(even_numbers) # [0, 2, 4, 6, 8]
```

```python
# 변수 = [표현식 if 조건 else 값 for 변수 in 반복가능객체]
# [참일_때_값 if 조건 else 거짓일_때_조건]
result = [x if x % 2 == 0 else 0 for x in range(1, 6)]
result = [x if x % 2 == 0 else 0 for x in range(1, 6)]
print(result)
```
