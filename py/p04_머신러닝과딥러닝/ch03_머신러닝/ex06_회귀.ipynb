```python
# 선형 회귀(Linear Regression)
# 선형 회귀 기반 모델 학습 및 예측
'''
문3-10> 주어진 코드를 사용하여 꽃받침의 길이와 너비, 꽃잎의 길이를 독립변수로 하여 꽃잎의 너비를 예측하여 선형 회귀 모델을 생성하고, 테스트셋을 사용하여 평가 지표를 확인하시오.
- 선형 회귀 모델의 가중치와 절편을 확인할 것
- 선형 회귀 모델의 평가 지표는 결정계수를 사용할 것
'''
# 패키지 로딩
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# 데이터 로딩
iris = load_iris()

# 로딩한 데이터를 데이터프레임으로 생성
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# 데이터 분할
y = iris_df['petal width (cm)']
X = iris_df.drop(['petal width (cm)'], axis=1)

# 학습용 데이터와 테스트용 데이터로 나눔
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
# 학습 모델1 - 선형 회귀 모델
lr_model = LinearRegression()

# 학습
lr_model.fit(X_train, y_train)
print('weight:', lr_model.coef_)
print('bias:', lr_model.intercept_)

# 평가
score = lr_model.score(X_test, y_test)
print('선형 회귀 모델 R2: {0:.2f}'.format(score))

# weight: [-0.24488106  0.24482729  0.53764071]
# bias: -0.14916391036571497
# 선형 회귀 모델 R2: 0.94
```

```python
# 트리 기반 회귀 모델 학습 및 예측
'''
문3-11> 문3-10의 데이터셋을 사용하여 의사결정나무를 사용한 회귀 모델을 생성하고 평가 지표를 확인하시오.
- 평가 지표는 결정계수를 사용, random_state=42로 설정할 것
'''
from sklearn.tree import DecisionTreeRegressor

 # 학습 - DecisionTreeRegressor
tree_model = DecisionTreeRegressor(random_state=42)
tree_model.fit(X_train, y_train)

# 평가
score = tree_model.score(X_test, y_test)
print('의사결정나무 R2: {0:.2f}'.format(score))
```

```python
'''
문3-12> 문3-10의 데이터셋을 사용하여 RandomForest, LGBM, XGB을 생성하여 학습을 진행하고, 가장 성능이 좋은 모델을 확인하시오.
- 평가 지표는 결정계수를 사용, random_state=42, n_estimators=100로 설정할 것
'''
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

# 학습
forest_model = RandomForestRegressor(random_state=42, n_estimators=100)
lgb_model = LGBMRegressor(random_state=42, n_estimators=100, verbosity=-1)
xgb_model = XGBRegressor(random_state=42, n_estimators=100)

model_list = [forest_model, lgb_model, xgb_model]

for model in model_list:
    model.fit(X_train, y_train)
    # 평가
    score = model.score(X_test, y_test)
    model_name = model.__class__.__name__
    print('{0} 모델 R2: {1:.2f}'.format(model_name, score))
```
