```python
# 교차 검증
'''
문5-1> 붓꽃 데이터를 KFold를 사용하여 3등분하고, 각 분할에서 데이터가 어떻게 분할되었는지 확인해보시오.
'''
import numpy as np
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold, StratifiedKFold, cross_validate
```

```python

# 데이터 불러오기
X, y = load_iris(return_X_y=True)
```

```python
# KFold로 데이터 분할하기
kf = KFold(n_splits=3)

# Fold는 1부터 시작
for Fold, (train, valid) in enumerate(kf.split(X), 1):
    print(f'---------Fold {Fold}---------')
    print('검증 데이터의 인덱스: {}~{}'.format(valid[0], valid[-1]))

    X_train, X_test = X[train], X[valid]
    y_train, y_test = y[train], y[valid]

    print('학습용 데이터 범주:', np.unique(y_train),
        '검증용 데이터 범주:', np.unique(y_test))
```

```python
# StratifiedKFold로 데이터 분할하기
'''
문5-2> 붓꽃 데이터를 StratifiedKFold를 사용하여 3등분하고, 각 분할에서 데이터가 어떻게 분할되었는지 확인해보시오.
'''
kf = StratifiedKFold(n_splits=3)
# Fold는 1부터 시작
# StratifiedKFold의 split() 메서드를 호출할 때 X와 y를 모두 할당해야 한다.
for Fold, (train, valid) in enumerate(kf.split(X, y), 1):
    print(f'---------Fold {Fold}---------')

    X_train, X_test = X[train], X[valid]
    y_train, y_test = y[train], y[valid]

    print('학습용 데이터 범주:', np.unique(y_train),
        '검증용 데이터 범주:', np.unique(y_test))
```

```python
# 교차검증 결과 확인하기
'''
문5-3> 문5-1에서 불러온 붓꽃 데이터를 사용하여 RandomForestClassifier를 생성하여 교차검증을 하려고 한다. 교차검증은 데이터셋을 3분할(3-Fold)하여 수행하고, 각 교차검증의 수행 결과를 출력하시오.
'''
model = RandomForestClassifier()

# 교차검증 수행
cv_results = cross_validate(model, X, y, cv=3)
print(cv_results)
```
