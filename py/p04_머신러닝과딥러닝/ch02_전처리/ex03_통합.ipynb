```python

import pandas as pd
```

```python
# 두 개의 통합할 데이터프레임
df1 = pd.DataFrame({
    'ID': ['1', '2', '3'],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': ['4', '5', '6'],
    'Name': ['David', 'Eva', 'Frank']
})
```

```python
df1
```

```python
df2
```

```python
# 데이터프레임 df1과 df2를 행(Row) 기준으로 연결(위아래로 붙이기)
pd.concat([df1, df2])
```

```python
# ignore_index=True -> 기존 인덱스 무시
# axis 기본값은 0(위에서 아래로 붙이기)
pd.concat([df1, df2], ignore_index=True)
```

```python
# 데이터 열 기준 결합
df3 = pd.DataFrame({
    'age': [25, 30, 35]
})
df3
```

```python
# 데이터프레임 df1과 df3를 열(Column) 기준으로 연결(좌우로 붙이기)
result_concat = pd.concat([df1, df3], axis=1)
print(result_concat)
```

```python
# 데이터 병합(Merge): inner(기본값), left, right, outer
df4 = pd.DataFrame({
    'ID': ['1', '2', '4', '5'],
    'Score': [85, 90, 95, 80]
})
```

```python
# 두 데이터프레임을 'ID'를 기준으로 병합 (기본: inner join)
result_merge = pd.merge(df1, df4, on='ID', how='inner')
print(result_merge)
```

```python
# 샘플 데이터프레임을 왼쪽 데이터셋(df1)의 'ID'를 키로 설정하여 병합
result_left = pd.merge(df1, df4, on='ID', how='left')
print(result_left)
```

```python
# 샘플 데이터프레임을 오른쪽 데이터셋(df4)의 'ID'를 키로 설정하여 병합
result_right = pd.merge(df1, df4, on='ID', how='right')
print(result_right)
```

```python
# 샘플 데이터프레임의 'ID'를 키로 설정하여 모든 데이터가 포함되도록 병합
result_outer = pd.merge(df1, df4, on='ID', how='outer')
print(result_outer)
```
