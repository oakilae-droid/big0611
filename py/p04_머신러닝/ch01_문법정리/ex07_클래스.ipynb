```python
# calculator.py
result = 0

def add(num):
    # global: result는 전역변수
    global result
    result += num  # result = result + num
    return result  # 결괏값 반환

print(add(3))
print(add(4))
```

```python
# calculator2.py
result1 = 0
result2 = 0

def add1(num):  # 계산기1
    global result1
    result1 += num
    return result1

def add2(num):  # 계산기2
    global result2
    result2 += num
    return result2

def sub1(num):  # 계산기3
    global result1
    result1 -= num
    return result1

def sub2(num):  # 계산기4
    global result2
    result2 -= num
    return result2

print(add1(3))
print(add1(4))
print(add2(3))
print(add2(7))
print(sub1(3))
print(sub1(4))
print(sub2(3))
print(sub2(7))
```

```python
# calculator3.py
# 클래스: 객체를 생성하기 위한 설계도(붕어빵 틀)
# 생성자 함수: 객체 생성시 초기값 설정 함수
# self: 객체 자기자신
class Calculator:
    # 객체가 생성될 때 자동으로 실행되는 초기화 함수
    def __init__(self):
        # cal1.result = 0
        # cal2.result = 0
        self.result = 0

    def add(self, num):
        # cal1.result = cal1.result + num
        # cal1.result = 0 + 3 -> 3
        # cal1.result = 3 + 4 -> 7
        self.result += num
        return self.result

    def sub(self, num):
        self.result -= num
        return self.result

# 객체(인스턴스): cal1, cal2
cal1 = Calculator()
cal2 = Calculator()

print(cal1.add(3))
print(cal1.add(4))
print(cal2.add(3))
print(cal2.add(7))
print(cal1.sub(3))
print(cal1.sub(4))
print(cal2.sub(3))
print(cal2.sub(7))
```

```python
# 클래스 정의
class FourCal:
    # 생성자 메서드
    # 객체가 생성되는 시점에 자동으로 호출된다
    def __init__(self, first, second):
        self.first = first
        self.second = second

    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        result = self.first + self.second
        return result

    def sub(self):
        result = self.first - self.second
        return result

    def mul(self):
        result = self.first * self.second
        return result
    
    def div(self):
        result = self.first / self.second
        return result
    
```

```python
# 객체 생성
# a는 객체이고, FourCal 클래스의 인스턴스이다.
# 바라보는 관점 차이
# 객체(Object): 구현할 수 있는 모든 실체(독립적인 대상)를 넓게 부르는 말
# 인스턴스(Instance): 특정 클래스와의 관계를 강조하여 부르는 말('어떤 클래스로 만들어졌는가?)
a = FourCal()
b = FourCal()
a
```

```python
# 상속(Inheritance)
# : 어떤 클래스(자식)를 만들 때 다른 클래스(부모)의 기능(함수->메서드)을 물려받을 수 있게 만드는 것
# class 클래스_이름(상속할_클래스_이름)
# 파이썬: **, pow(2, 3)
# 자바스크립트: **(ES7 이후), Math.pow(2, 3)
# 자바: ** 없다, Math.pow(2, 3)
class MoreFourCal(FourCal):
    def usepow(self):
        result = self.first ** self.second
        return result

a = MoreFourCal(4, 2)
print(a.add())
print(a.sub())
print(a.mul())
print(a.div())
print(a.usepow())

pow(2, 10)
```

```python
class SafeFourCal(FourCal):
    # 메서드 재정의(method overriding)
    # : 부모 클래스(상속한 클래스)에 있는 메서드를 동일한 이름으로 다시 만드는 것
    # 부모 클래스의 메서드 대신 오버라이딩한 메서드가 호출된다.
    def div(self):
        if self.second == 0:  # 나누는 값이 0인 경우 0을 반환하도록 수정
            return 0
        else:
            return self.first / self.second

a = SafeFourCal(4, 0)
a.div()
```

```python
# 클래스변수
# : 객체변수와 달리 클래스로 만든 모든 객체에 공유된다.
class Family:
    # 클래스 변수: lastname
    lastname = "박"

    # 객체 변수: first, second
    def setdata(self, first, second):
        self.first = first
        self.second = second  

# 클래스_이름.클래스변수로 사용
Family.lastname      

a = Family()
b = Family()
print(a.lastname)
print(b.lastname)
```
