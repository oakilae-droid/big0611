```python
# EDA(Exploratory Data Analysis, 탐색적 데이터 분석)
# : 데이터의 날것 그대로의 상태를 샅샅이 뜯어보며 특징을 파악하고 인사이트를 얻는 과정

import pandas as pd

data = {
    "고객ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    "성별": ["남", "여", "여", "남", "여", "남", "남", "여", "남", "여"],
    "나이": [25, 32, 28, 41, 35, 22, 45, 30, 38, 27],
    "지역": ["서울", "부산", "서울", "대구", "인천", "서울", "부산", "대전", "서울", "인천"],
    "구매금액": [35000, 72000, 15000, 120000, 55000, 23000, 98000, 45000, 87000, 32000],
    "구매횟수": [2, 5, 1, 8, 4, 2, 6, 3, 5, 2],
    "회원등급": ["일반", "VIP", "일반", "VVIP", "VIP", "일반", "VIP", "일반", "VIP", "일반"],
    "만족도": [4, 5, 3, 5, 4, 2, 5, 4, 4, 3]
}

df = pd.DataFrame(data)
```

```python
# 1단계: 데이터의 첫인상 확인하기 (구조 파악)
df.head()       # 상위 5개 데이터
```

```python
df.shape        # 행, 열 크기
```

```python
df.columns      # 열 이름
```

```python
# 2단계: 데이터의 건강 상태 체크하기 (결측치와 타입)
df.info()               # 각 열의 데이터 타입과 Non-Null 개수 확인
```

```python
df.isnull().sum()       # 결측치가 열마다 몇 개씩 있는지 누적 합계 확인
```

```python
# 3단계: 요약 통계로 수치 훑어보기 (기술 통계)
df.describe()           # 수치형 데이터의 요약 통계량
```

```python
df['성별'].value_counts()  # 범주형 데이터의 종류별 개수 (예: 남성 5명, 여성 5명)
```

```python
# 4단계: 시각화로 숨은 패턴 찾기 (시각적 탐색)
# - 숫자만 봐서는 모르는 데이터의 '분포'와 변수 간의 '관계'를 그래프로 시각화합니다.

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 1. 단일 변수 탐색: 고객들의 나이 분포는 어떨까? (히스토그램)
sns.histplot(df['나이'], kde=True)
plt.show()
```

```python
# 2. 다변수 탐색: 나이와 구매금액은 어떤 관계가 있을까? (산점도)
sns.scatterplot(x='나이', y='구매금액', data=df)
plt.show()
```
